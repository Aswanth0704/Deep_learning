```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def linear_function1D(x, beta, omega):
  y = beta + omega*x
  return y

x =  np.arange(0.0, 10.0, 0.01)
beta = 10.0
omega = -2.0

y = linear_function1D(x, beta, omega)

# plot
fig, ax = plt.subplots()
ax.plot(x, y, 'r-')
ax.set_ylim([0, 10]), ax.set_xlim([0, 10])
ax.set_xlabel('x')
ax.set_ylabel('y')
plt.show()

```

```python
from matplotlib import contour
def draw_2D_function(x1_mesh, x2_mesh, y):
  fig, ax = plt.subplots()
  fig.set_size_inches(7, 7)
  pos = ax.contourf(x1_mesh, x2_mesh, y, levels = 256, cmap = 'hot', vmin = -10.0, vmax = 10.0)
  fig.colorbar(pos, ax = ax)
  ax.set_xlabel('x1')
  ax.set_ylabel('x2')
  # levels = np.arange(-10, 10, 1.0)
  # ax.contourf(x1_mesh, x2_mesh, y, levels, cmap = 'winter')
  plt.show()
```

```python
# linear function with two inputs x1, x2:
def linear_function_2D(x1, x2, beta, omega1, omega2):
  y = beta + omega1*x1 + omega2*x2
  return y
```

```python
# Plot 2D function
x1 = np.arange(0.0, 10.0, 0.1)
x2 = np.arange(0.0, 10.0, 0.1)
x1, x2 = np.meshgrid(x1, x2) # x1: 100 x 100; x2: 100 x 100

print(x1.shape)
print(x2.shape)

# Compute the 2D function
beta = 0.0; omega1 = 1.0; omega2 = -0.5
y = linear_function_2D(x1, x2, beta, omega1, omega2)

# draw function
draw_2D_function(x1, x2, y)
```

```python
## Use the Matrix multiplication using

# Define the parameters
beta1 = 0.5; beta2 = 0.2
omega11 =  -1.0 ; omega12 = 0.4; omega13 = -0.3
omega21 =  0.1  ; omega22 = 0.1; omega23 = 1.2

x1 = 4.0; x2 = -1; x3 = 2.0

beta_vec = np.array([[beta1], [beta2]]) # 2x1;
omega_mat = np.array([[omega11, omega12, omega13], [omega21, omega22, omega23]]) # 2x3
x_vec = np.array([[x1], [x2], [x3]]) # 3x1

# Compute with vectors/matrix
y_vec = beta_vec + omega_mat @ x_vec

print(y_vec)
```
