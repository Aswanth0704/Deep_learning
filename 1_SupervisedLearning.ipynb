```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([0.03, 0.19, 0.34, 0.46, 0.78, 0.81, 1.08, 1.18, 1.39, 1.60, 1.65, 1.90])
y = np.array([0.67, 0.85, 1.05, 1.0, 1.40, 1.5, 1.3, 1.54, 1.55, 1.68, 1.73, 1.6])

print(x)
print(y)
```

```python
# Linear representation of y_hat = phi0 + phi1*x in 1D
def f(x, phi0, phi1):
  y = phi0 + phi1*x
  return y

def plot_fun(x, y, phi0, phi1):
  fig, ax = plt.subplots()
  ax.scatter(x, y)
  ax.set_xlim([0, 2.0])
  ax.set_ylim([0, 2.0])
  ax.set_xlabel('x')
  ax.set_ylabel('y')

  # draw a line
  x_line = np.arange(0, 2, 0.01)
  y_line = f(x_line, phi0, phi1)
  plt.plot(x_line, y_line, 'b-', lw = 2)
  plt.show()

phi0 = 0.4
phi1 = 0.2

plot_fun(x, y, phi0, phi1)
```

```python
# loss function
def compute_loss(x, y, phi0, phi1):
  y_hat = f(x, phi0, phi1)
  loss = np.mean((y-y_hat)**2)
  return loss

loss = compute_loss(x, y, phi0, phi1)
print(f"loss: {loss}")
```

```python
phi0 = 1.6; phi1 = -0.8
loss = compute_loss(x, y, phi0, phi1)
print(f"loss: {loss}")
plot_fun(x, y, phi0, phi1)
```

```python
phi0 = 0.7; phi1 = 0.5
loss = compute_loss(x, y, phi0, phi1)
print(f"loss: {loss}")
plot_fun(x, y, phi0, phi1)
```
